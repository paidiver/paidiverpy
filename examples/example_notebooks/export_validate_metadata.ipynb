## Open config file and metadata

This example shows how to handle and parse metadata and configuration files using paidiverpy package

## Import dependencies

```python
%load_ext autoreload
%autoreload 2
from paidiverpy.pipeline import Pipeline
```

```python
pipeline = Pipeline(config_file_path="../config_files/config_benthic.yml")
pipeline.run()
```

```python
pipeline.get_metadata().shape
```

```python
pipeline.metadata.metadata.shape
```

```python
pipeline.metadata.export_metadata("csv")
```

```python
pipeline.metadata.export_metadata("json")
```

```python
pipeline.metadata.export_metadata("ifdo")
```
